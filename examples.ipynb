```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.spatial import ConvexHull

# 1. Կետեր և ուռուցիկ թաղանթ
np.random.seed(5)
points = np.random.uniform(10, 90, size=(20, 2))
hull = ConvexHull(points)

# 2. Ստուգում ենք q կետը 
q = np.array([55, 65])
inside = np.all(hull.equations[:, :-1] @ q + hull.equations[:, -1] <= 1e-12)

# 3. Նկարում ենք
plt.figure(figsize=(6.5, 5))
plt.scatter(points[:, 0], points[:, 1], label='Կետեր')
for s in hull.simplices:
    plt.plot(points[s, 0], points[s, 1], 'orange', lw=2)
plt.scatter(q[0], q[1], color='green' if inside else 'red', s=100, zorder=5, label='Ստուգիչ կետ')
plt.title(f'Կետը {"ներսում է" if inside else "դրսում է"}')
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.spatial import ConvexHull

# 1. 20 պատահական կետ [0, 10]x[0, 10] քառակուսում (մակերեսը = 100) և ուռուցիկ թաղանթը
np.random.seed(3)
pts = np.random.uniform(0, 10, size=(20, 2))
hull = ConvexHull(pts)

# 2. Թաղանթի մակերեսը (2D-ում hull.volume-ն մակերեսն է) և զբաղեցրած մասը (%)
area = hull.volume
ratio = (area / 100.0) * 100

# 3. Նկարում ենք
plt.figure(figsize=(6, 5))
plt.scatter(pts[:, 0], pts[:, 1])
for s in hull.simplices:
    plt.plot(pts[s, 0], pts[s, 1], 'orange', lw=2)
plt.fill(pts[hull.vertices, 0], pts[hull.vertices, 1], 'orange', alpha=0.2)
plt.title(f'Թաղանթի մակերեսը՝ {area:.1f} ({ratio:.1f}% քառակուսուց)')
plt.xlim(0, 10); plt.ylim(0, 10)
plt.show()
```
